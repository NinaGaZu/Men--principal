Función seek() para manipular la posición

```python
archivo = open("archivo.txt", "r")

# define offset and desde_donde (whence) before calling seek
offset = 0  # number of bytes to move
desde_donde = 0  # 0=start of file, 1=current position, 2=end of file
```

Ejemplos de aplicación:

```python
archivo.seek(offset, desde_donde)

# Mover el puntero al inicio del archivo
archivo.seek(0)

# Mover el puntero al byte 50 desde el inicio
archivo.seek(50, 0)

# Mover el puntero 10 bytes adelante desde la posición actual
archivo.seek(10, 1)

# Mover el puntero al final del archivo
archivo.seek(0, 2)
```

Función tell() para conocer la posición actual

```python
posicion_actual = archivo.tell()
print(f"El puntero está en la posición: {posicion_actual}")
```

Utilidad en operaciones específicas

```python
with open('datos.txt', 'r') as archivo:
    # Leer los primeros 100 bytes
    primer_bloque = archivo.read(100)
    
    # Mover el puntero al byte 200
    archivo.seek(200)
    
    # Leer los siguientes 50 bytes desde esa posición
    segundo_bloque = archivo.read(50)
    
    # Verificar posición actual
    print(f"Posición actual: {archivo.tell()}")
```

Módulo Pickle

```python
objeto = {'nombre': 'Juan', 'edad': 30, 'ciudad': 'Madrid'}


import pickle

# Guardar objeto
with open('datos.pkl', 'wb') as archivo:
    pickle.dump(objeto, archivo)

# Cargar objeto
with open('datos.pkl', 'rb') as archivo:
    objeto = pickle.load(archivo)
```
